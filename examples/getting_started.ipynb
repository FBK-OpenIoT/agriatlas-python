Set the collection and the item to open, and where to save the files.

```python
COLLECTION_ID = "<collection-id>"
ITEM_ID = "<item-id>"
```

## Imports

```python
import agriatlas
```

## Log in

`login()` asks for your username and password, unless `AGRIATLAS_USERNAME` and `AGRIATLAS_PASSWORD` are set. It returns an authenticated STAC client. If you want to set directly the credentials, call `login(<username>, <password>)`.

```python
catalog = agriatlas.login()
print(agriatlas.get_session())
```

## Open the item

```python
item = catalog.get_collection(COLLECTION_ID).get_item(ITEM_ID)

print("Title:", item.properties.get("title", item.id))
print("Description:", item.properties.get("description", "-"))
print("Assets:")
for key, asset in item.assets.items():
    size = asset.extra_fields.get("file:size")
    print(f"  {key}: {asset.title or ''} roles={asset.roles} size={size}")
```

## Download the data

Downloads the assets with the `data` role into `DATA_DIR`. Files already downloaded are skipped, and checksums are verified.

```python
DATA_DIR = "./data"
files = agriatlas.download(item, DATA_DIR)
files
```
